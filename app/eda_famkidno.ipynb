# EDA: FAMKIDNO

FAMKIDNO = number of family members under 18 (0, 1, 2, 3 where 3 means 3 or more).

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the dataframe that was saved in extraction.ipynb
# (makes df available / "global" across notebooks)
df = pd.read_parquet("nhis_data.parquet")
df.shape
```

```python
# Quick summary of the variable
df["FAMKIDNO"].describe()
```

```python
# How many people fall in each category
counts = df["FAMKIDNO"].value_counts().sort_index()
counts
```

```python
# Bar chart of the distribution of FAMKIDNO
plt.figure(figsize=(7, 5))
plt.bar(counts.index.astype(str), counts.values, color="steelblue")
plt.xlabel("Number of family members under 18 (FAMKIDNO)")
plt.ylabel("Number of people")
plt.title("Distribution of FAMKIDNO")

# add the count on top of each bar
for x, y in zip(counts.index.astype(str), counts.values):
    plt.text(x, y, str(y), ha="center", va="bottom")

plt.show()
```

```python
# Same thing as percentages, which is easier to interpret
pct = df["FAMKIDNO"].value_counts(normalize=True).sort_index() * 100

plt.figure(figsize=(7, 5))
plt.bar(pct.index.astype(str), pct.values, color="darkorange")
plt.xlabel("Number of family members under 18 (FAMKIDNO)")
plt.ylabel("Percent of people (%)")
plt.title("Distribution of FAMKIDNO (percent)")

for x, y in zip(pct.index.astype(str), pct.values):
    plt.text(x, y, f"{y:.1f}%", ha="center", va="bottom")

plt.show()
```
