```python
import pandas as pd
import matplotlib.pyplot as plt
import set_fig_theme
```

```python
df = pd.read_parquet("nhis_data.parquet")
print(df.shape)
print(df.head())
```

```python
df['URBRRL'].value_counts()
```

```python
urbrrl_cats = [1, 2, 3, 4]
urbrrl_cats_names = ['Large Central Metro', 'Large fringe metro', 'Medium and small metro', 'Nonmetropolitan']

df['URBRRL'].value_counts(normalize = True).reindex(urbrrl_cats).plot.barh(color = 'darkorange', edgecolor = 'black')
plt.yticks([0, 1, 2, 3], urbrrl_cats_names, rotation = 45)
plt.ylabel('Urban-Rural County Classification (URBRRL)')
plt.xlabel('Fraction of Sample')
plt.title("Urban-Rural County Classification Across \nNHIS Participants' Households (2024)")
```
