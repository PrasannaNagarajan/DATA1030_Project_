```python
import pandas as pd
import matplotlib.pyplot as plt
import set_fig_theme
```

```python
df = pd.read_parquet("nhis_data.parquet")
print(df.shape)
print(df.head())
```

```python
total_values = df['URBRRL'].value_counts()
print(total_values)

total = len(df['URBRRL'])
percentages = [(v / total) * 100 for v in total_values]
print(percentages)
```

```python
urbrrl_cats = [1, 2, 3, 4]
urbrrl_cats_names = ['Large Central Metro', 'Large fringe metro', 'Medium and small metro', 'Nonmetropolitan']

ax = df['URBRRL'].value_counts(normalize = True).reindex(urbrrl_cats).mul(100).plot.barh(color = 'darkorange', edgecolor = 'black')
ax.bar_label(ax.containers[0], fmt = '%.1f%%', padding = 1)
plt.yticks([0, 1, 2, 3], urbrrl_cats_names, rotation = 45)
plt.xlim(0, 35)
plt.ylabel('Urban-Rural County Classification (URBRRL)')
plt.xlabel('Percent of Sample (%)')
plt.title("Urban-Rural County Classification Across \nNHIS Participants' Households (2024)")
```
